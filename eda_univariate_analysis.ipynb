## 1. Import Libraries


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2. Load Dataset


```python
df = pd.read_csv('train.csv')
df
```

# 3. Categorical Data

Categorical data contains categories or labels such as **Sex** and **Embarked**.

## A. Countplot

A countplot shows how many observations belong to each category.

```python
sns.countplot(x=df['Embarked'])
plt.show()
```

```python
sns.countplot(x = df['Survived'])
plt.show()
df['Survived'].value_counts()
```

## B. Pie Chart

A pie chart shows the proportion of each category.

```python
df['Survived'].value_counts().plot(kind='pie', autopct='%.2f')
plt.ylabel('')
plt.show()
```

# 4. Numerical Data

Numerical data contains numbers, such as **Age**.

## A. Histogram

A histogram shows the distribution of numerical values.

```python
plt.hist(df['Age'].dropna(), bins=50)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

## B. Distribution Plot

The original notebook uses `sns.distplot()`. That function is deprecated in modern Seaborn, so `histplot(..., kde=True)` is used here for the same general purpose.

```python
sns.histplot(df['Age'].dropna(), kde=True)
plt.show()
```

## C. Boxplot

A boxplot helps visualize the spread of numerical data and identify possible outliers.

```python
sns.boxplot(x=df['Age'])
plt.show()
```

## 5. Basic Age Statistics


```python
print('Minimum:', df['Age'].min())
print('Maximum:', df['Age'].max())
print('Mean:', df['Age'].mean())
print('Skewness:', df['Age'].skew())
```

## Quick Revision

| Visualization | Used for |
|---|---|
| `countplot()` | Count of categorical values |
| `pie` | Proportion of categories |
| `hist()` | Distribution of numerical data |
| `histplot()` | Distribution + optional KDE |
| `boxplot()` | Spread and outliers |
| `min()` | Minimum value |
| `max()` | Maximum value |
| `mean()` | Average value |
| `skew()` | Skewness of distribution |
